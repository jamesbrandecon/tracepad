%%ai 
Load data/retail_orders.csv, parse order_date, and return a useful preview of the orders.

```python
import pandas as pd

tracepad_result_1 = pd.read_csv('data/retail_orders.csv', parse_dates=['order_date'])

orders = tracepad_result_1  # Tracepad result: @orders points to this same object
try:
    from tracepad import present
except ImportError:
    present = lambda value, **_: value
present(orders, name="orders")
```

%%ai 
Using @orders, summarize monthly net revenue and order count by channel

```python
# Tracepad references: friendly names point to the same objects; no data is copied
orders = tracepad_result_1  # @orders

orders_with_month = orders.assign(
    order_month=orders["order_date"].dt.to_period("M").dt.to_timestamp(),
    net_revenue=orders["revenue"] * (1 - orders["returned"])
)
tracepad_result_2 = (
    orders_with_month.groupby(["order_month", "channel"], as_index=False)
    .agg(net_revenue=("net_revenue", "sum"), order_count=("order_id", "nunique"))
    .sort_values(["order_month", "channel"])
)

monthly_revenue = tracepad_result_2  # Tracepad result: @monthly_revenue points to this same object
from tracepad import present
present(monthly_revenue, name="monthly_revenue")
```

> **Follow-up 2.1** · using `@monthly_revenue`
>
> Using @monthly_revenue, plot net revenue over time with one line per channel.

```python
# Tracepad references: friendly names point to the same objects; no data is copied
monthly_revenue = tracepad_result_2  # @monthly_revenue

import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(9, 4.5))
for channel, values in monthly_revenue.groupby("channel"):
    ax.plot(values["order_month"], values["net_revenue"], marker="o", label=channel)
ax.set(title="Monthly net revenue by channel", xlabel="Month", ylabel="Net revenue")
ax.legend(frameon=False)
fig.tight_layout()
tracepad_result_2_1 = fig

revenue_chart = tracepad_result_2_1  # Tracepad result: @revenue_chart points to this same object
from tracepad import present
present(revenue_chart, name="revenue_chart")
```
